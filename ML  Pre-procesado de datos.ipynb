La variable dependiente es la que el algoritmo va a intentar predecir a partir de las otras variables, en este caso purchased, las independientes son las variables que le suministramos al algoritmo para que intente predecir 

### Transfromar la variable dependiente, training y testing, dividiendo el dataset

```python
# ---Limpieza de NaN por su nombre de columna ---
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())


# --- Convertir a Dummies por nombre de columna ---
# Convertimos 'Country' a dummies y 'Purchased' a 0 y 1 de forma explícita
df = pd.get_dummies(df, columns=['Country'], dtype=int)
df['Purchased'] = df['Purchased'].map({'Yes': 1, 'No': 0})

# --- Separar en X (independientes) e y (dependiente) ---
# Aquí los datos siguen teniendo nombres legibles en Pandas
X = df.drop(columns=['Purchased'])
y = df['Purchased']



# --- División en Entrenamiento y Prueba ---
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)



# --- Escalado óptimo por NOMBRE DE COLUMNA (Sin usar [:, 3:]) ---
sc = StandardScaler()

# Escalamos únicamente la edad y el salario usando sus nombres directamente
X_train[['Age', 'Salary']] = sc.fit_transform(X_train[['Age', 'Salary']])
X_test[['Age', 'Salary']] = sc.transform(X_test[['Age', 'Salary']])



print("Mostrar X_train escalada")
print(X_train)
print("Mostrar X_test escalada")
X_test

```
